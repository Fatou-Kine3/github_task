```python
def MSE(y_pred, y):
    """Calculate mean squared error.

    Parameters
    ----------
    y_pred : ndarray of shape (n_samples,)
        Predicted values.
    y : ndarray of shape (n_samples,)
        True target values.

    Returns
    -------
    mse : float
        Mean squared error.
    """

    # Check that both arrays have the same shape
    if y_pred.shape != y.shape:
        raise ValueError("y_pred and y must have the same shape.")

    # Calculate mean squared error
    return np.mean((y_pred - y) ** 2)
```

```python
def fit(self, X, y, X_val=None, y_val=None):
    """Train the linear regression model."""

    # Add intercept column when bias is used
    if not self.no_bias:
        X = np.column_stack([np.ones(X.shape[0]), X])
        if X_val is not None:
            X_val = np.column_stack([np.ones(X_val.shape[0]), X_val])

    # Initialize parameters
    self.theta = np.zeros(X.shape[1])

    for i in range(self.iter):

        # Calculate predictions
        y_pred = self._linear_hypothesis(X)

        # Calculate error
        error = y_pred - y

        # Update parameters using training data only
        self._gradient_descent(X, error)

        # Calculate objective function
        y_pred = self._linear_hypothesis(X)
        self.loss[i] = 0.5 * MSE(y_pred, y)

        # Calculate validation loss if validation data is provided
        if X_val is not None and y_val is not None:
            y_val_pred = self._linear_hypothesis(X_val)
            self.val_loss[i] = 0.5 * MSE(y_val_pred, y_val)

        if self.verbose:
            print(f"Iteration {i + 1}/{self.iter}")
```

```python
import numpy as np


class ScratchLinearRegression:
    """Scratch implementation of linear regression."""

    def __init__(self, num_iter, lr, no_bias, verbose):
        self.iter = num_iter
        self.lr = lr
        self.no_bias = no_bias
        self.verbose = verbose
        self.theta = np.array([])
        self.loss = np.zeros(self.iter)
        self.val_loss = np.zeros(self.iter)

    def _linear_hypothesis(self, X):
        """Calculate predictions from the current parameters."""
        return X @ self.theta

    def _gradient_descent(self, X, error):
        """Update the parameters by one gradient descent step."""

        # Number of training samples
        m = X.shape[0]

        # Calculate the gradient
        gradient = self.lr / m * X.T @ error

        # Update the parameters
        self.theta -= gradient

    def fit(self, X, y, X_val=None, y_val=None):
        """Train the linear regression model."""

        # Add an intercept column when bias is used
        if not self.no_bias:
            X = np.column_stack([np.ones(X.shape[0]), X])

            if X_val is not None:
                X_val = np.column_stack([np.ones(X_val.shape[0]), X_val])

        # Initialize parameters
        self.theta = np.zeros(X.shape[1])

        for i in range(self.iter):

            # Calculate predictions
            y_pred = self._linear_hypothesis(X)

            # Calculate error
            error = y_pred - y

            # Update parameters using training data
            self._gradient_descent(X, error)

            # Calculate training objective
            y_pred = self._linear_hypothesis(X)
            self.loss[i] = 0.5 * MSE(y_pred, y)

            # Calculate validation objective
            if X_val is not None and y_val is not None:
                y_val_pred = self._linear_hypothesis(X_val)
                self.val_loss[i] = 0.5 * MSE(y_val_pred, y_val)

            if self.verbose:
                print(f"Iteration {i + 1}/{self.iter}")

    def predict(self, X):
        """Predict target values."""

        # Add the intercept column in the same way as during training
        if not self.no_bias:
            X = np.column_stack([np.ones(X.shape[0]), X])

        return self._linear_hypothesis(X)


def MSE(y_pred, y):
    """Calculate mean squared error."""

    # Check that both arrays have the same shape
    if y_pred.shape != y.shape:
        raise ValueError("y_pred and y must have the same shape.")

    return np.mean((y_pred - y) ** 2)
```

```python
## Problem 10 — Derivation of the Gradient Descent Update

The objective function is:

J(θ) = (1/2m) Σᵢ(hθ(x⁽ⁱ⁾) - y⁽ⁱ⁾)²

where the linear hypothesis is:

hθ(x⁽ⁱ⁾) = Σⱼ θⱼxⱼ⁽ⁱ⁾

We want to calculate the partial derivative of J(θ) with respect
to one parameter θⱼ.

### Step 1: Differentiate the objective function

∂J(θ)/∂θⱼ
= (1/2m) Σᵢ 2(hθ(x⁽ⁱ⁾) - y⁽ⁱ⁾)
  ∂(hθ(x⁽ⁱ⁾) - y⁽ⁱ⁾)/∂θⱼ

The 2 cancels with 1/2:

∂J(θ)/∂θⱼ
= (1/m) Σᵢ (hθ(x⁽ⁱ⁾) - y⁽ⁱ⁾)
  ∂hθ(x⁽ⁱ⁾)/∂θⱼ

### Step 2: Differentiate the hypothesis

Since:

hθ(x⁽ⁱ⁾) = Σⱼ θⱼxⱼ⁽ⁱ⁾

we have:

∂hθ(x⁽ⁱ⁾)/∂θⱼ = xⱼ⁽ⁱ⁾

Therefore:

∂J(θ)/∂θⱼ
= (1/m) Σᵢ
  (hθ(x⁽ⁱ⁾) - y⁽ⁱ⁾)xⱼ⁽ⁱ⁾

### Step 3: Apply gradient descent

The gradient descent rule is:

θⱼ ← θⱼ - α ∂J(θ)/∂θⱼ

Substituting the derivative:

θⱼ ← θⱼ
- (α/m) Σᵢ
  (hθ(x⁽ⁱ⁾) - y⁽ⁱ⁾)xⱼ⁽ⁱ⁾

This is the update equation used in Problem 2.

### Vector form

For all parameters simultaneously:

θ ← θ - (α/m) Xᵀ(Xθ - y)

In the implementation, this corresponds to:

gradient = self.lr / m * X.T @ error

self.theta -= gradient
```

```python
## Problem 11 — Local Optima

The objective function of linear regression is:

J(θ) = (1/2m) ||Xθ - y||²

This function is convex because its Hessian matrix is:

H = (1/m) XᵀX

For any vector z:

zᵀHz = (1/m) zᵀXᵀXz
     = (1/m) ||Xz||² ≥ 0

Therefore, the Hessian is positive semidefinite and the objective
function is convex.

A convex function has no suboptimal local minima. If a minimum exists,
it is also a global minimum.

For a simple one-parameter example, the objective function has a
bowl-shaped form:

J(θ) = (1/2m) Σᵢ(θx⁽ⁱ⁾ - y⁽ⁱ⁾)²

The gradient points toward the global minimum.

### Learning rate

The learning rate controls the size of each gradient descent step.

If the learning rate is too small, convergence is very slow.

If the learning rate is too large, the algorithm can overshoot the
minimum and fail to converge. It may oscillate or even diverge.

Therefore, an appropriate learning rate is necessary for gradient
descent to converge efficiently.
```

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(-5, 5, 200)

# A simple convex quadratic objective function
J = (theta - 1) ** 2

plt.figure(figsize=(8, 5))
plt.plot(theta, J, linewidth=2)

plt.xlabel("Theta")
plt.ylabel("Objective function J(theta)")
plt.title("Convex Objective Function of Linear Regression")

plt.grid()
plt.show()
```
